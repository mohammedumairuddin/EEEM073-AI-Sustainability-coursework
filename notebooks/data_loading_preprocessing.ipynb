```python
import pandas as pd

df = pd.read_csv("crop_production.csv")
df.head()
```

```python
print("=== DATASET STATISTICS (copy these into your report) ===")
print(f"Raw records:       {df.shape[0]:,}")
print(f"Raw columns:       {df.shape[1]}")
print(f"Missing values:\n{df.isnull().sum()}")
```

```python
#creating yield features
import numpy as np
df["Yield"] = df["Production"] / df["Area"]
df = df.replace([np.inf, -np.inf], np.nan)
```

```python
df = df.dropna() #defining dropna to remove missing values
```

```python
print(f"Records after cleaning: {df.shape[0]:,}")
print(f"Records removed:        removed rows logged above")
```

```python
#encoding categorical variables like state name, seasons, crops.
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

categorical_columns = ["State_Name","District_Name","Season","Crop"]

for col in categorical_columns:
    df[col] = le.fit_transform(df[col])
```

```python
import os

# Create the 'data' directory if it doesn't already exist
os.makedirs('data', exist_ok=True)
df.to_csv('data/processed_dataset.csv', index=False)
```
