```python
import pandas as pd

df = pd.read_csv("crop_production.csv")
df.head()
```

```python
#printing the records and values
print(f"Raw records:       {df.shape[0]:,}")
print(f"Raw columns:       {df.shape[1]}")
print(f"Missing values:\n{df.isnull().sum()}")
```

```python
#creating yield features
import numpy as np
df["Yield"] = df["Production"] / df["Area"]
df = df.replace([np.inf, -np.inf], np.nan)
```

```python
df = df.dropna() #defining dropna to remove missing values
```

```python
print(f"Records after cleaning: {df.shape[0]:,}")
print(f"Records removed:        removed rows logged above")
```

```python
#encoding categorical variables like state name, seasons, crops.
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

categorical_columns = ["State_Name","District_Name","Season","Crop"]

for col in categorical_columns:
    df[col] = le.fit_transform(df[col])
```

```python
import os

# Creating the 'data' directory
os.makedirs('data', exist_ok=True)
df.to_csv('data/processed_dataset.csv', index=False)
```

```python
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"Zero-area records removed: {(df['Area'] == 0).sum()}")
print(f"Negative yield records: {(df['Yield'] < 0).sum()}")
print(f"Yield — mean: {df['Yield'].mean():.2f}, median: {df['Yield'].median():.2f}")
print(f"Yield — std:  {df['Yield'].std():.2f}")
print(f"Yield — skewness: {df['Yield'].skew():.4f}")
print(f"Yield — kurtosis: {df['Yield'].kurt():.4f}")
```
